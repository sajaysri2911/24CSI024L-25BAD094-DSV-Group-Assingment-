### PROBLEM STATEMENT :
#### KidneyCare Analytics: Early Detection of Chronic Kidney Disease Using Clinical Health Indicators

### Domain:
#### Healthcare & Public Health

### Project Abstract:
Chronic Kidney Disease (CKD) is a major global health concern that may remain undetected until it reaches advanced stages. This project aims to analyze clinical health indicators and identify the key factors associated with Chronic Kidney Disease using Data Science and Visualization techniques.

The project will begin with raw clinical data and perform a systematic analysis to understand the structure, quality, and characteristics of the dataset. The raw dataset will later be preprocessed by handling missing values, cleaning inconsistent data, detecting and treating outliers, and transforming relevant features.

Exploratory Data Analysis (EDA), statistical analysis, Principal Component Analysis (PCA), and interactive visualizations will be used to identify hidden patterns and relationships among clinical health indicators. The final objective is to generate meaningful insights that can support the understanding of factors associated with Chronic Kidney Disease and contribute to data-driven healthcare decision-making.

### Dataset Source:

**Source:** Kaggle

**Dataset Link:** https://www.kaggle.com/datasets/mansoordaku/ckdisease
