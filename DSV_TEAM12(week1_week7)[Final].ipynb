## **7. Handle Numerical Missing Values**

**Purpose:** Replace missing numerical values using the median.

**Reason:** The median is less affected by extreme values.

```python
for col in numerical_cols:
    if df_cleaned[col].isnull().sum() > 0:
        df_cleaned[col] = df_cleaned[col].fillna(
            df_cleaned[col].median()
        )
```

## **8. Handle Categorical Missing Values**

**Purpose:** Replace missing categorical values using the mode.

**Reason:** The mode represents the most frequently occurring category.

```python
for col in categorical_cols:
    if df_cleaned[col].isnull().sum() > 0:
        df_cleaned[col] = df_cleaned[col].fillna(
            df_cleaned[col].mode()[0]
        )
```

## **12. Outlier Detection Using Z-Score**

**Purpose:** Identify observations that significantly deviate from the mean.

A Z-score greater than 3 or less than -3 is considered a potential outlier.

```python
all_numerical_cols = numerical_cols

z_score_df = pd.DataFrame(index=df_cleaned.index)

for col in all_numerical_cols:
    mean = df_cleaned[col].mean()
    std = df_cleaned[col].std()

    z_score_df[col] = (
        df_cleaned[col] - mean
    ) / std

z_outlier_mask = z_score_df.abs() > 3

z_outlier_report = pd.DataFrame({
    "Outlier Count": z_outlier_mask.sum(),
    "Outlier Percentage": (
        z_outlier_mask.sum() / len(df_cleaned)
    ) * 100
})

z_outlier_report
```

## **2. Numerical Distribution Analysis Before Scaling:**
**Purpose**: Examine the distribution and range of the selected numerical features before applying scaling.

**Explanation**: Histograms help identify skewness, spread, and differences in the ranges of clinical measurements. Features with large numerical values may dominate features with smaller values in some machine learning algorithms.

## **3.Standardization Using StandardScaler:**
**Purpose:** Standardize numerical features so that each feature has a mean approximately equal to 0 and a standard deviation approximately equal to 1.

**Explanation:** StandardScaler transforms the values based on their mean and standard deviation. It is useful when numerical features have different units and ranges. However, it does not remove outliers, and extreme values can still influence the results.

```python
from sklearn.preprocessing import StandardScaler

# Create the StandardScaler
standard_scaler = StandardScaler()

# Apply standardization
standard_scaled_data = standard_scaler.fit_transform(
    df_cleaned[numerical_features]
)

# Convert the result into a DataFrame
df_standard_scaled = pd.DataFrame(
    standard_scaled_data,
    columns=numerical_features,
    index=df_cleaned.index
)

print("Dataset After StandardScaler:")
display(df_standard_scaled.head())

print("\nStatistics After StandardScaler:")
display(df_standard_scaled.describe().round(2))
```

## **4.Normalization Using MinMaxScaler:**
**Purpose:** Transform numerical features into a common range, usually between 0 and 1.

**Explanation**: MinMaxScaler maps the minimum value of each feature to 0 and the maximum value to 1. This makes features easier to compare on a common scale. However, the transformation is sensitive to extreme values because they influence the minimum and maximum.

```python
from sklearn.preprocessing import MinMaxScaler

# Create the MinMaxScaler
minmax_scaler = MinMaxScaler()

# Apply normalization
minmax_scaled_data = minmax_scaler.fit_transform(
    df_cleaned[numerical_features]
)

# Convert the result into a DataFrame
df_minmax_scaled = pd.DataFrame(
    minmax_scaled_data,
    columns=numerical_features,
    index=df_cleaned.index
)

print("Dataset After MinMaxScaler:")
display(df_minmax_scaled.head())

print("\nMinimum and Maximum After MinMaxScaler:")
display(
    pd.DataFrame({
        "Minimum": df_minmax_scaled.min(),
        "Maximum": df_minmax_scaled.max()
    }).round(2)
)
```

## **5. Before-and-After Scaling Comparison:**
**Purpose:** Compare the original numerical values with the values obtained using StandardScaler and MinMaxScaler.

**Explanation:** The comparison shows how each technique changes the numerical range. StandardScaler produces values centered around zero, whereas MinMaxScaler transforms the minimum and maximum values to 0 and 1, respectively.

```python
# Compare the first five rows of selected features
comparison = pd.concat([
    df_cleaned[numerical_features].head().add_suffix("_Before"),
    df_standard_scaled.head().add_suffix("_Standard"),
    df_minmax_scaled.head().add_suffix("_MinMax")
], axis=1)

print("Before and After Scaling Comparison:")
display(comparison.round(3))
```

## **6. Visual Comparison of Before and After Scaling:**
**Purpose:** Visualize the original and scaled distributions of numerical features.

**Explanation:** These plots demonstrate how scaling changes the numerical range of the features. The underlying observations remain in the same order, but their numerical values are transformed according to the selected scaling method.

```python
# Compare the distributions of the first four numerical features
features_to_plot = numerical_features[:4]

for col in features_to_plot:
    plt.figure(figsize=(8, 4))

    sns.kdeplot(
        df_cleaned[col],
        label="Before Scaling"
    )

    sns.kdeplot(
        df_standard_scaled[col],
        label="StandardScaler"
    )

    sns.kdeplot(
        df_minmax_scaled[col],
        label="MinMaxScaler"
    )

    plt.title("Scaling Comparison: " + col)
    plt.xlabel("Feature Value")
    plt.ylabel("Density")
    plt.legend()
    plt.show()
```

## **7. Final Scaled Numerical Datasets:**

## **9. Categorical Encoding and Skewness Transformation**

### **Objective:**
Identify categorical variables in the CKD dataset, encode them using suitable techniques, analyze skewness in numerical features, apply transformations where needed, and compare distributions before and after transformation.

## **10. Identify Categorical Columns**

The following CKD variables are treated as categorical when present in the dataset: `rbc`, `pc`, `pcc`, `ba`, `htn`, `dm`, `cad`, `appet`, `pe`, `ane`, and `classification`.

```python
# Identify requested categorical columns in the cleaned CKD dataset
requested_categorical_columns = [
    "rbc", "pc", "pcc", "ba", "htn",
    "dm", "cad", "appet", "pe", "ane",
    "classification"
]

ckd_categorical_columns = [
    col for col in requested_categorical_columns
    if col in df_cleaned.columns
]

print("Categorical columns found:")
print(ckd_categorical_columns)

for col in ckd_categorical_columns:
    print(f"\n{col} - unique values:")
    print(df_cleaned[col].value_counts(dropna=False))
```

## **11. Encode Categorical Variables**

Binary predictor variables are encoded as 0/1, while predictors with more than two categories are one-hot encoded. The `classification` target is encoded separately and is not included in predictor encoding.

```python
# Create a separate encoded copy; df_cleaned remains unchanged
df_encoded = df_cleaned.copy()

# Normalize whitespace in categorical string values
for col in ckd_categorical_columns:
    df_encoded[col] = df_encoded[col].apply(
        lambda value: value.strip() if isinstance(value, str) else value
    )

target_column = "classification"

# Encode the target separately
if target_column in df_encoded.columns:
    target_values = sorted(df_encoded[target_column].dropna().unique())
    target_mapping = {value: index for index, value in enumerate(target_values)}
    df_encoded[target_column] = df_encoded[target_column].map(target_mapping)
    print("Classification target mapping:", target_mapping)

predictor_categorical_columns = [
    col for col in ckd_categorical_columns
    if col != target_column
]

binary_categorical_columns = [
    col for col in predictor_categorical_columns
    if df_encoded[col].nunique(dropna=True) == 2
]

multicategory_columns = [
    col for col in predictor_categorical_columns
    if df_encoded[col].nunique(dropna=True) > 2
]

# Binary encoding for two-category predictors
for col in binary_categorical_columns:
    values = sorted(df_encoded[col].dropna().unique())
    mapping = {value: index for index, value in enumerate(values)}
    df_encoded[col] = df_encoded[col].map(mapping)
    print(f"{col} mapping:", mapping)

# One-hot encoding for predictors with more than two categories
if multicategory_columns:
    df_encoded = pd.get_dummies(
        df_encoded,
        columns=multicategory_columns,
        dtype=int
    )

print("\nEncoded dataset preview:")
display(df_encoded.head())
print("Encoded dataset shape:", df_encoded.shape)
```

## **12. Check Skewness of Numerical Variables**

Skewness measures distribution asymmetry. Values above 1 or below -1 are treated as potential signs of high skewness; transformation decisions are then checked against the actual feature distributions.

```python
# Select numerical clinical features; exclude identifier and target columns
ckd_numerical_columns = df_cleaned.select_dtypes(include=np.number).columns.tolist()
ckd_numerical_columns = [
    col for col in ckd_numerical_columns
    if col.lower() not in ["id", "classification"]
]

# Calculate skewness
ckd_skewness_before = df_cleaned[ckd_numerical_columns].skew()

ckd_skewness_report = pd.DataFrame({
    "Feature": ckd_skewness_before.index,
    "Skewness": ckd_skewness_before.values
})

ckd_skewness_report["Interpretation"] = ckd_skewness_report["Skewness"].apply(
    lambda value: "Highly positively skewed" if value > 1
    else "Highly negatively skewed" if value < -1
    else "Approximately symmetric or moderately skewed"
)

ckd_skewness_report = ckd_skewness_report.sort_values(
    by="Skewness",
    key=lambda values: values.abs(),
    ascending=False
)

display(ckd_skewness_report.round(3))
```

## **14. Apply Transformation Only Where Needed**

The Yeo–Johnson power transformation can handle positive, zero, and negative values. It is applied only to features whose absolute skewness is greater than 1 and which have enough distinct values to make transformation meaningful. The original `df_cleaned` data is preserved.

```python
from sklearn.preprocessing import PowerTransformer

# Preserve the original cleaned dataset
df_ckd_transformed = df_cleaned.copy()

# Select only highly skewed numerical features with sufficient variation
ckd_features_to_transform = [
    col for col in ckd_numerical_columns
    if pd.notna(ckd_skewness_before[col])
    and abs(ckd_skewness_before[col]) > 1
    and df_cleaned[col].dropna().nunique() > 2
]

print("Features selected for transformation:")
print(ckd_features_to_transform)

# Transform observed values only, preserving any missing values
for col in ckd_features_to_transform:
    valid_mask = df_ckd_transformed[col].notna()
    observed_values = df_ckd_transformed.loc[valid_mask, [col]]

    transformer = PowerTransformer(
        method="yeo-johnson",
        standardize=False
    )

    df_ckd_transformed.loc[valid_mask, col] = (
        transformer.fit_transform(observed_values).ravel()
    )

print("Selective transformation completed.")
print("Number of transformed features:", len(ckd_features_to_transform))
```

## **15. Compare Skewness Before and After Transformation**

The comparison table shows whether the absolute skewness decreased or increased for each numerical feature. A positive reduction means absolute skewness decreased; a negative reduction means it increased.

```python
ckd_skewness_after = df_ckd_transformed[ckd_numerical_columns].skew()

ckd_skewness_comparison = pd.DataFrame({
    "Feature": ckd_numerical_columns,
    "Skewness Before": [
        ckd_skewness_before[col] for col in ckd_numerical_columns
    ],
    "Skewness After": [
        ckd_skewness_after[col] for col in ckd_numerical_columns
    ]
})

ckd_skewness_comparison["Absolute Skewness Reduction"] = (
    ckd_skewness_comparison["Skewness Before"].abs()
    - ckd_skewness_comparison["Skewness After"].abs()
)

display(
    ckd_skewness_comparison.sort_values(
        by="Absolute Skewness Reduction",
        ascending=False
    ).round(3)
)
```

The paired histograms provide visual evidence of how each selected feature's distribution changed after transformation.

## **WEEK 7:**
### **FEATURE SELECTION**

### **Objective:**

The objective of this week is to identify the most useful features from the engineered Chronic Kidney Disease (CKD) dataset prepared during Week 6. Feature selection helps remove irrelevant and redundant features, reduce dataset complexity, and retain important clinical indicators for predicting kidney disease.

The following techniques are used:

* Variance-based feature selection
* Correlation-based feature analysis
* Statistical feature selection using SelectKBest

The final output is a reduced dataset containing the selected features and the target variable.



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.impute import SimpleImputer

# Use the engineered dataset from Week 6
df_feature_selection = df_encoded.copy()

# Remove the identifier column
if "id" in df_feature_selection.columns:
    df_feature_selection = df_feature_selection.drop(columns=["id"])

# Separate features and target
X = df_feature_selection.drop(columns=["classification"])
y = df_feature_selection["classification"]

# Convert input features to numeric format
X = X.apply(pd.to_numeric, errors="coerce")

# Remove columns containing only missing values
X = X.dropna(axis=1, how="all")

# Replace infinite values with missing values
X = X.replace([np.inf, -np.inf], np.nan)

# Fill missing values using the median
imputer = SimpleImputer(strategy="median")

X = pd.DataFrame(
    imputer.fit_transform(X),
    columns=X.columns,
    index=X.index
)

print("Dataset shape:", df_feature_selection.shape)
print("Number of input features:", X.shape[1])
print("Target variable: classification")
```

## **1. Variance-Based Feature Selection**

Variance-based feature selection identifies features with little or no variation across observations. Features with zero variance have the same value for every patient and generally provide no useful information for distinguishing between patients.

The variance of each feature is calculated, and constant features are identified. These features are removed from the dataset, while the remaining features are retained for further analysis.

```python
# Calculate variance for each feature

feature_variance = X.var()

variance_report = pd.DataFrame({
    "Feature": feature_variance.index,
    "Variance": feature_variance.values
})

variance_report = variance_report.sort_values("Variance")

display(variance_report)

# Identify constant features
constant_features = feature_variance[
    feature_variance == 0
].index.tolist()

print("Constant features:", constant_features)

# Remove constant features
X_variance = X.drop(columns=constant_features)

print("Features before selection:", X.shape[1])
print("Features after selection:", X_variance.shape[1])
```

## **2. Correlation-Based Feature Analysis**

Correlation analysis is performed to understand the relationships between numerical features in the CKD dataset. A correlation matrix and heatmap are used to visualize the strength and direction of relationships between variables.

Highly correlated features may contain similar information and could introduce redundancy. Feature pairs with an absolute correlation greater than 0.90 are identified for further examination.

The results help determine which features may be redundant. However, highly correlated features are not automatically removed because their clinical relevance and relationship with the target variable must also be considered.

```python
# Calculate the correlation matrix

correlation_matrix = X_variance.corr()

# Visualize the correlation matrix
plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap of CKD Features")
plt.tight_layout()
plt.show()

# Identify highly correlated feature pairs
upper_triangle = correlation_matrix.where(
    np.triu(
        np.ones(correlation_matrix.shape),
        k=1
    ).astype(bool)
)

correlation_threshold = 0.90

high_correlation_pairs = []

for column in upper_triangle.columns:
    for row in upper_triangle.index:
        value = upper_triangle.loc[row, column]

        if pd.notna(value) and abs(value) > correlation_threshold:
            high_correlation_pairs.append(
                (row, column, value)
            )

print("Highly correlated feature pairs:")

for feature1, feature2, value in high_correlation_pairs:
    print(f"{feature1} - {feature2}: {value:.3f}")

if not high_correlation_pairs:
    print("No highly correlated feature pairs found.")
```

## **3. Statistical Feature Selection Using SelectKBest**

Statistical feature selection is used to identify features that have a useful statistical relationship with the CKD classification target.

The SelectKBest method with the ANOVA F-test is applied to rank the input features. Features with higher F-scores show stronger evidence of differences between the target classes.

The top 10 features, or all available features if fewer than 10 remain, are selected for the reduced dataset. The target variable is kept separate from the input features during selection.

```python
# Encode target labels for statistical selection

y_encoded = pd.factorize(y)[0]

# Select up to 10 features
k = min(10, X_variance.shape[1])

selector = SelectKBest(
    score_func=f_classif,
    k=k
)

selector.fit(X_variance, y_encoded)

# Create feature-ranking report
feature_scores = pd.DataFrame({
    "Feature": X_variance.columns,
    "F_Score": selector.scores_,
    "P_Value": selector.pvalues_
})

feature_scores = feature_scores.sort_values(
    by="F_Score",
    ascending=False
)

print("Feature Ranking:")
display(feature_scores.round(4))

# Get selected features
selected_features = X_variance.columns[
    selector.get_support()
].tolist()

print("Selected features:")
print(selected_features)
```

## **4. Creation of the Reduced Dataset**

After completing variance analysis, correlation analysis, and statistical feature selection, the selected features are used to create a reduced dataset.

The reduced dataset contains the selected clinical indicators and the `classification` target variable. Features that were not selected are documented as removed features.

The resulting dataset is saved as a CSV file for use in subsequent project stages. The actual selected and removed features are determined from the analysis results.

```python
# Create the reduced dataset

df_reduced = X_variance[selected_features].copy()

# Add the target variable
df_reduced["classification"] = y

# Identify selected and removed features
removed_features = [
    feature
    for feature in X.columns
    if feature not in selected_features
]

print("Selected features:")
print(selected_features)

print("\nRemoved features:")
print(removed_features)

print("\nOriginal dataset shape:", df_encoded.shape)
print("Reduced dataset shape:", df_reduced.shape)

# Display the reduced dataset
display(df_reduced.head())

# Save the reduced dataset
df_reduced.to_csv(
    "ckd_feature_selected.csv",
    index=False
)

print("Reduced dataset saved successfully.")
```

### **5. Results and Observations**

The CKD dataset is analyzed using variance-based, correlation-based, and statistical feature selection techniques.

- Variance analysis identifies constant features that can be removed.
- Correlation analysis reveals relationships and possible redundancy between features.
- SelectKBest ranks features according to their statistical relationship with the classification target.
- The identifier column is excluded from predictive features.
- The selected features are used to create a reduced dataset.
- The selected and removed feature lists are recorded for documentation.

### **Conclusion**

Feature selection helps identify useful clinical indicators and reduce unnecessary features in the CKD dataset. The selected features and target variable are saved in `ckd_feature_selected.csv`. This reduced dataset is prepared for subsequent dimensionality reduction and machine learning tasks.

## **PRINCIPAL COMPONENT ANALYSIS (PCA)**

### **Objective**

PCA is applied to the selected CKD features to reduce dimensionality while retaining as much information as possible. The selected features are standardized before PCA because PCA is sensitive to differences in feature scale.

### **1. Load the Selected Features**

The feature-selected dataset created in the previous section is used as the input for PCA. The `classification` column is kept separately as the target and is not included in the PCA calculation.

```python
df_feature_selected = pd.read_csv("ckd_feature_selected.csv")

X_pca = df_feature_selected.drop(columns=["classification"])
y_pca = df_feature_selected["classification"]

print("Selected dataset shape:", df_feature_selected.shape)
print("PCA input shape:", X_pca.shape)
print("Target shape:", y_pca.shape)

display(df_feature_selected.head())
```

### **2. Standardize the Selected Features**

PCA is sensitive to feature scale. Standardization gives each selected feature a mean of approximately 0 and a standard deviation of 1, preventing features with larger numerical ranges from dominating the principal components.

```python
pca_scaler = StandardScaler()

X_pca_scaled = pca_scaler.fit_transform(X_pca)

X_pca_scaled = pd.DataFrame(
    X_pca_scaled,
    columns=X_pca.columns,
    index=X_pca.index
)

print("Standardized PCA input:")
display(X_pca_scaled.head())
```

### **3. Apply Principal Component Analysis**

PCA transforms the standardized features into a smaller set of uncorrelated principal components. Each component represents a combination of the original features and captures a portion of the total variance in the dataset.

```python
from sklearn.decomposition import PCA

pca = PCA()

X_pca_transformed = pca.fit_transform(X_pca_scaled)

print("Original number of features:", X_pca_scaled.shape[1])
print("Number of principal components:", X_pca_transformed.shape[1])
```

### **4. Explained Variance**

Explained variance shows how much of the total information in the standardized dataset is captured by each principal component. Components with larger explained variance contribute more to representing the original data.

```python
explained_variance = pca.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

pca_variance_report = pd.DataFrame({
    "Principal Component": [
        f"PC{i}" for i in range(1, len(explained_variance) + 1)
    ],
    "Explained Variance Ratio": explained_variance,
    "Explained Variance (%)": explained_variance * 100,
    "Cumulative Variance (%)": cumulative_variance * 100
})

display(pca_variance_report.round(4))
```

### **5. Explained Variance by Principal Component**

The explained variance plot shows the contribution of each principal component. It helps identify how quickly the information content decreases across the components.

```python
plt.figure(figsize=(10, 5))

plt.plot(
    range(1, len(explained_variance) + 1),
    explained_variance,
    marker="o"
)

plt.xlabel("Principal Component")
plt.ylabel("Explained Variance Ratio")
plt.title("Explained Variance by Principal Component")
plt.xticks(range(1, len(explained_variance) + 1))
plt.grid(True)
plt.tight_layout()
plt.show()
```

### **6. Cumulative Explained Variance**

Cumulative explained variance shows how much total information is retained when multiple principal components are combined. It is used to determine an appropriate number of components for dimensionality reduction.

```python
plt.figure(figsize=(10, 5))

plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    marker="o"
)

plt.axhline(
    y=0.90,
    linestyle="--",
    label="90% Variance"
)

plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Cumulative Explained Variance")
plt.xticks(range(1, len(cumulative_variance) + 1))
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

### **7. Selecting the Number of Principal Components**

The smallest number of principal components that retains at least 90% of the total variance is selected. This provides a practical balance between dimensionality reduction and information retention.

```python
components_for_90 = np.argmax(cumulative_variance >= 0.90) + 1

variance_retained_90 = cumulative_variance[
    components_for_90 - 1
] * 100

print("Components required for at least 90% variance:", components_for_90)
print("Variance retained:", round(variance_retained_90, 2), "%")
```

### **8. Create the Reduced PCA Dataset**

The selected number of principal components is used to create the final reduced dataset. The original classification target is added back after PCA so that the reduced representation can still be related to the CKD classification.

```python
pca_reduced = PCA(
    n_components=components_for_90
)

X_reduced = pca_reduced.fit_transform(X_pca_scaled)

pca_columns = [
    f"PC{i}" for i in range(1, components_for_90 + 1)
]

df_pca = pd.DataFrame(
    X_reduced,
    columns=pca_columns,
    index=X_pca.index
)

df_pca["classification"] = y_pca.values

print("Final PCA dataset shape:", df_pca.shape)

display(df_pca.head())
```
