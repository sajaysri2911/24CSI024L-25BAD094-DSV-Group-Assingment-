# **Week-3:**
#Data Cleaning:

**Objective:** Clean the dataset by fixing inconsistent values, correcting data types, handling missing values, checking duplicates, and detecting potential outliers.

# 1. Initial Data Quality Check

**Purpose:** Record the condition of the dataset before cleaning.


```python
df_cleaned = df.copy()
print("Dataset Shape:", df_cleaned.shape)
print("Total Missing Values:", df_cleaned.isnull().sum().sum())
print("Duplicate Records:", df_cleaned.duplicated().sum())
```

# 2. Clean Categorical Values

**Purpose:** Remove unnecessary spaces and tab characters from categorical values.

```python
categorical_columns = [
    "rbc", "pc", "pcc", "ba",
    "htn", "dm", "cad",
    "appet", "pe", "ane",
    "classification"
]

for col in categorical_columns:
    df_cleaned[col] = (
        df_cleaned[col]
        .astype("string")
        .str.strip()
        .str.lower()
    )
```

# 3. Verify Categorical Values

**Purpose:** Check whether categorical values are consistently formatted.

```python
for col in categorical_columns:
    print(f"\n{col}:")
    print(df_cleaned[col].unique())
```

# 4. Convert Incorrect Data Types

**Purpose:** Convert numerical values stored as text into numeric data types.

```python
numeric_object_columns = ["pcv", "wc", "rc"]

for col in numeric_object_columns:
    df_cleaned[col] = pd.to_numeric(
        df_cleaned[col],
        errors="coerce"
    )

print(df_cleaned[numeric_object_columns].dtypes)
```

# 5. Check Missing Values

**Purpose:** Identify missing values after correcting data types.

```python
missing_summary = pd.DataFrame({
    "Missing Values": df_cleaned.isnull().sum(),
    "Missing Percentage": (
        df_cleaned.isnull().sum() / len(df_cleaned) * 100
    ).round(2)
})

missing_summary
```

# 6. Identify Feature Types

**Purpose:** Separate numerical and categorical features for appropriate data cleaning.

```python
numerical_cols = df_cleaned.select_dtypes(
    include=np.number
).columns.tolist()

categorical_cols = df_cleaned.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

# Remove identifier column
numerical_cols.remove("id")

print("Numerical Features:")
print(numerical_cols)

print("\nCategorical Features:")
print(categorical_cols)
```

# 7. Handle Numerical Missing Values

**Purpose:** Replace missing numerical values using the median.

**Reason:** The median is less affected by extreme values.

```python
for col in numerical_cols:
    if df_cleaned[col].isnull().sum() > 0:
        df_cleaned[col] = df_cleaned[col].fillna(
            df_cleaned[col].median()
        )
```

# 8. Handle Categorical Missing Values

**Purpose:** Replace missing categorical values using the mode.

**Reason:** The mode represents the most frequently occurring category.

```python
for col in categorical_cols:
    if df_cleaned[col].isnull().sum() > 0:
        df_cleaned[col] = df_cleaned[col].fillna(
            df_cleaned[col].mode()[0]
        )
```

# 9. Verify Missing Values

**Purpose:** Confirm that missing values have been handled successfully.

```python
print("Total Missing Values:")
print(df_cleaned.isnull().sum().sum())

print("\nMissing Values by Column:")
print(df_cleaned.isnull().sum())
```

# 10. Update Feature Lists

**Purpose:** Confirm the numerical and categorical features before outlier detection.

```python
numerical_cols = df_cleaned.select_dtypes(
    include=np.number
).columns.tolist()

categorical_cols = df_cleaned.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

numerical_cols.remove("id")

print("Numerical Features:")
print(numerical_cols)

print("\nCategorical Features:")
print(categorical_cols)

df_cleaned.to_csv("cleaned_dataset.csv",index=False)
```

# 11. Outlier Detection Using Boxplots

**Purpose:** Visually identify potential outliers in numerical clinical features.

```python
for col in numerical_cols:
    plt.figure(figsize=(8, 4))

    df_cleaned.boxplot(column=col)

    plt.title(f"Boxplot of {col}")
    plt.ylabel(col)

    plt.show()
```

# 12. Outlier Detection Using Z-Score

**Purpose:** Identify observations that significantly deviate from the mean.

A Z-score greater than 3 or less than -3 is considered a potential outlier.

```python
all_numerical_cols = numerical_cols

z_score_df = pd.DataFrame(index=df_cleaned.index)

for col in all_numerical_cols:
    mean = df_cleaned[col].mean()
    std = df_cleaned[col].std()

    z_score_df[col] = (
        df_cleaned[col] - mean
    ) / std

z_outlier_mask = z_score_df.abs() > 3

z_outlier_report = pd.DataFrame({
    "Outlier Count": z_outlier_mask.sum(),
    "Outlier Percentage": (
        z_outlier_mask.sum() / len(df_cleaned)
    ) * 100
})

z_outlier_report
```

# **Week-4:**
#Exploratory Data Analysis (EDA)

**Objective:** Explore the cleaned kidney disease dataset using univariate and bivariate analysis. The analysis focuses on distributions, category frequencies, relationships between clinical variables, and differences between kidney disease classes.

**Note:** The cleaned dataset from Week 3 is used for analysis. No rows are removed in this stage; EDA is used to understand patterns and generate hypotheses for later feature engineering and dimensionality reduction.

```python
# Week 4 setup
import matplotlib.pyplot as plt
import seaborn as sns

# Use the cleaned dataframe produced in Week 3
eda_df = df_cleaned.copy()

# Make the target column easier to read in plots
eda_df['classification'] = eda_df['classification'].astype(str).str.strip().str.lower()

print('EDA dataset shape:', eda_df.shape)
eda_df.head()
```

## 1. Target Variable Distribution

The target variable is `classification`. Its distribution is checked to understand class balance before comparing groups.

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=eda_df, x='classification')
plt.title('Distribution of Kidney Disease Classification')
plt.xlabel('Classification')
plt.ylabel('Number of Patients')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print(eda_df['classification'].value_counts())
print('\nPercentage distribution:')
print((eda_df['classification'].value_counts(normalize=True) * 100).round(2))
```

## 2. Univariate Analysis: Numerical Features

Histograms with KDE curves are used to study the shape, spread, and possible skewness of important numerical clinical variables.

```python
numeric_eda_cols = [col for col in numerical_cols if col != 'id']

for col in numeric_eda_cols:
    plt.figure(figsize=(7, 4))
    sns.histplot(data=eda_df, x=col, kde=True, bins=20)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.tight_layout()
    plt.show()
```

## 3. Univariate Analysis: Categorical Features

Count plots show the frequency of categories. These plots help identify dominant categories and whether rare categories may affect later encoding or modelling.

```python
categorical_eda_cols = [col for col in categorical_cols if col != 'classification']

for col in categorical_eda_cols:
    plt.figure(figsize=(7, 4))
    order = eda_df[col].value_counts().index
    sns.countplot(data=eda_df, x=col, order=order)
    plt.title(f'Category Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

## 4. Bivariate Analysis: Numerical Features vs Classification

Boxplots compare numerical clinical measurements across the two target classes. Differences in median and spread can indicate variables worth investigating further.

```python
for col in numeric_eda_cols:
    plt.figure(figsize=(7, 4))
    sns.boxplot(data=eda_df, x='classification', y=col)
    plt.title(f'{col} by Kidney Disease Classification')
    plt.xlabel('Classification')
    plt.ylabel(col)
    plt.tight_layout()
    plt.show()
```

## 5. Bivariate Analysis: Categorical Features vs Classification

Stacked percentage tables and heatmaps are used to compare categorical features with the target. Percentages are calculated within each feature category to make comparisons easier.

```python
for col in categorical_eda_cols:
    crosstab = pd.crosstab(eda_df[col], eda_df['classification'], normalize='index') * 100
    print(f'\n{col} vs classification (%)')
    display(crosstab.round(2))

    crosstab.plot(kind='bar', stacked=True, figsize=(7, 4))
    plt.title(f'{col} vs Classification')
    plt.xlabel(col)
    plt.ylabel('Percentage within category')
    plt.xticks(rotation=45)
    plt.legend(title='Classification')
    plt.tight_layout()
    plt.show()
```

## 6. Correlation Analysis

The correlation matrix is limited to numerical variables. It helps identify relationships and possible multicollinearity for the Week 6–7 feature engineering and PCA stages. Correlation does not establish causation.

```python
corr_cols = [col for col in numeric_eda_cols if eda_df[col].nunique() > 1]
correlation_matrix = eda_df[corr_cols].corr()

plt.figure(figsize=(12, 9))
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Matrix of Numerical Features')
plt.tight_layout()
plt.show()
```
