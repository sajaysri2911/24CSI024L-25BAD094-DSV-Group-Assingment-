# 7. Handle Numerical Missing Values

**Purpose:** Replace missing numerical values using the median.

**Reason:** The median is less affected by extreme values.

```python
for col in numerical_cols:
    if df_cleaned[col].isnull().sum() > 0:
        df_cleaned[col] = df_cleaned[col].fillna(
            df_cleaned[col].median()
        )
```

# 8. Handle Categorical Missing Values

**Purpose:** Replace missing categorical values using the mode.

**Reason:** The mode represents the most frequently occurring category.

```python
for col in categorical_cols:
    if df_cleaned[col].isnull().sum() > 0:
        df_cleaned[col] = df_cleaned[col].fillna(
            df_cleaned[col].mode()[0]
        )
```

# 12. Outlier Detection Using Z-Score

**Purpose:** Identify observations that significantly deviate from the mean.

A Z-score greater than 3 or less than -3 is considered a potential outlier.

```python
all_numerical_cols = numerical_cols

z_score_df = pd.DataFrame(index=df_cleaned.index)

for col in all_numerical_cols:
    mean = df_cleaned[col].mean()
    std = df_cleaned[col].std()

    z_score_df[col] = (
        df_cleaned[col] - mean
    ) / std

z_outlier_mask = z_score_df.abs() > 3

z_outlier_report = pd.DataFrame({
    "Outlier Count": z_outlier_mask.sum(),
    "Outlier Percentage": (
        z_outlier_mask.sum() / len(df_cleaned)
    ) * 100
})

z_outlier_report
```

## 3.Standardization Using StandardScaler:
**Purpose:** Standardize numerical features so that each feature has a mean approximately equal to 0 and a standard deviation approximately equal to 1.

**Explanation:** StandardScaler transforms the values based on their mean and standard deviation. It is useful when numerical features have different units and ranges. However, it does not remove outliers, and extreme values can still influence the results.

```python
from sklearn.preprocessing import StandardScaler

# Create the StandardScaler
standard_scaler = StandardScaler()

# Apply standardization
standard_scaled_data = standard_scaler.fit_transform(
    df_cleaned[numerical_features]
)

# Convert the result into a DataFrame
df_standard_scaled = pd.DataFrame(
    standard_scaled_data,
    columns=numerical_features,
    index=df_cleaned.index
)

print("Dataset After StandardScaler:")
display(df_standard_scaled.head())

print("\nStatistics After StandardScaler:")
display(df_standard_scaled.describe().round(2))
```

## 4.Normalization Using MinMaxScaler:
**Purpose:** Transform numerical features into a common range, usually between 0 and 1.

**Explanation**: MinMaxScaler maps the minimum value of each feature to 0 and the maximum value to 1. This makes features easier to compare on a common scale. However, the transformation is sensitive to extreme values because they influence the minimum and maximum.

```python
from sklearn.preprocessing import MinMaxScaler

# Create the MinMaxScaler
minmax_scaler = MinMaxScaler()

# Apply normalization
minmax_scaled_data = minmax_scaler.fit_transform(
    df_cleaned[numerical_features]
)

# Convert the result into a DataFrame
df_minmax_scaled = pd.DataFrame(
    minmax_scaled_data,
    columns=numerical_features,
    index=df_cleaned.index
)

print("Dataset After MinMaxScaler:")
display(df_minmax_scaled.head())

print("\nMinimum and Maximum After MinMaxScaler:")
display(
    pd.DataFrame({
        "Minimum": df_minmax_scaled.min(),
        "Maximum": df_minmax_scaled.max()
    }).round(2)
)
```

# 9. Categorical Encoding and Skewness Transformation

## Objective
Identify categorical variables in the CKD dataset, encode them using suitable techniques, analyze skewness in numerical features, apply transformations where needed, and compare distributions before and after transformation.

## 9.1 Identify Categorical Columns

The following CKD variables are treated as categorical when present in the dataset: `rbc`, `pc`, `pcc`, `ba`, `htn`, `dm`, `cad`, `appet`, `pe`, `ane`, and `classification`.

```python
# Identify requested categorical columns in the cleaned CKD dataset
requested_categorical_columns = [
    "rbc", "pc", "pcc", "ba", "htn",
    "dm", "cad", "appet", "pe", "ane",
    "classification"
]

ckd_categorical_columns = [
    col for col in requested_categorical_columns
    if col in df_cleaned.columns
]

print("Categorical columns found:")
print(ckd_categorical_columns)

for col in ckd_categorical_columns:
    print(f"\n{col} - unique values:")
    print(df_cleaned[col].value_counts(dropna=False))
```

## 9.2 Encode Categorical Variables

Binary predictor variables are encoded as 0/1, while predictors with more than two categories are one-hot encoded. The `classification` target is encoded separately and is not included in predictor encoding.

```python
# Create a separate encoded copy; df_cleaned remains unchanged
df_encoded = df_cleaned.copy()

# Normalize whitespace in categorical string values
for col in ckd_categorical_columns:
    df_encoded[col] = df_encoded[col].apply(
        lambda value: value.strip() if isinstance(value, str) else value
    )

target_column = "classification"

# Encode the target separately
if target_column in df_encoded.columns:
    target_values = sorted(df_encoded[target_column].dropna().unique())
    target_mapping = {value: index for index, value in enumerate(target_values)}
    df_encoded[target_column] = df_encoded[target_column].map(target_mapping)
    print("Classification target mapping:", target_mapping)

predictor_categorical_columns = [
    col for col in ckd_categorical_columns
    if col != target_column
]

binary_categorical_columns = [
    col for col in predictor_categorical_columns
    if df_encoded[col].nunique(dropna=True) == 2
]

multicategory_columns = [
    col for col in predictor_categorical_columns
    if df_encoded[col].nunique(dropna=True) > 2
]

# Binary encoding for two-category predictors
for col in binary_categorical_columns:
    values = sorted(df_encoded[col].dropna().unique())
    mapping = {value: index for index, value in enumerate(values)}
    df_encoded[col] = df_encoded[col].map(mapping)
    print(f"{col} mapping:", mapping)

# One-hot encoding for predictors with more than two categories
if multicategory_columns:
    df_encoded = pd.get_dummies(
        df_encoded,
        columns=multicategory_columns,
        dtype=int
    )

print("\nEncoded dataset preview:")
display(df_encoded.head())
print("Encoded dataset shape:", df_encoded.shape)
```

## 9.3 Check Skewness of Numerical Variables

Skewness measures distribution asymmetry. Values above 1 or below -1 are treated as potential signs of high skewness; transformation decisions are then checked against the actual feature distributions.

```python
# Select numerical clinical features; exclude identifier and target columns
ckd_numerical_columns = df_cleaned.select_dtypes(include=np.number).columns.tolist()
ckd_numerical_columns = [
    col for col in ckd_numerical_columns
    if col.lower() not in ["id", "classification"]
]

# Calculate skewness
ckd_skewness_before = df_cleaned[ckd_numerical_columns].skew()

ckd_skewness_report = pd.DataFrame({
    "Feature": ckd_skewness_before.index,
    "Skewness": ckd_skewness_before.values
})

ckd_skewness_report["Interpretation"] = ckd_skewness_report["Skewness"].apply(
    lambda value: "Highly positively skewed" if value > 1
    else "Highly negatively skewed" if value < -1
    else "Approximately symmetric or moderately skewed"
)

ckd_skewness_report = ckd_skewness_report.sort_values(
    by="Skewness",
    key=lambda values: values.abs(),
    ascending=False
)

display(ckd_skewness_report.round(3))
```

## 9.5 Apply Transformation Only Where Needed

The Yeo–Johnson power transformation can handle positive, zero, and negative values. It is applied only to features whose absolute skewness is greater than 1 and which have enough distinct values to make transformation meaningful. The original `df_cleaned` data is preserved.

```python
from sklearn.preprocessing import PowerTransformer

# Preserve the original cleaned dataset
df_ckd_transformed = df_cleaned.copy()

# Select only highly skewed numerical features with sufficient variation
ckd_features_to_transform = [
    col for col in ckd_numerical_columns
    if pd.notna(ckd_skewness_before[col])
    and abs(ckd_skewness_before[col]) > 1
    and df_cleaned[col].dropna().nunique() > 2
]

print("Features selected for transformation:")
print(ckd_features_to_transform)

# Transform observed values only, preserving any missing values
for col in ckd_features_to_transform:
    valid_mask = df_ckd_transformed[col].notna()
    observed_values = df_ckd_transformed.loc[valid_mask, [col]]

    transformer = PowerTransformer(
        method="yeo-johnson",
        standardize=False
    )

    df_ckd_transformed.loc[valid_mask, col] = (
        transformer.fit_transform(observed_values).ravel()
    )

print("Selective transformation completed.")
print("Number of transformed features:", len(ckd_features_to_transform))
```

## 9.6 Compare Skewness Before and After Transformation

The comparison table shows whether the absolute skewness decreased or increased for each numerical feature. A positive reduction means absolute skewness decreased; a negative reduction means it increased.

```python
ckd_skewness_after = df_ckd_transformed[ckd_numerical_columns].skew()

ckd_skewness_comparison = pd.DataFrame({
    "Feature": ckd_numerical_columns,
    "Skewness Before": [
        ckd_skewness_before[col] for col in ckd_numerical_columns
    ],
    "Skewness After": [
        ckd_skewness_after[col] for col in ckd_numerical_columns
    ]
})

ckd_skewness_comparison["Absolute Skewness Reduction"] = (
    ckd_skewness_comparison["Skewness Before"].abs()
    - ckd_skewness_comparison["Skewness After"].abs()
)

display(
    ckd_skewness_comparison.sort_values(
        by="Absolute Skewness Reduction",
        ascending=False
    ).round(3)
)
```

## 9.7 Compare Distributions Before and After Transformation

The paired histograms provide visual evidence of how each selected feature's distribution changed after transformation.

```python
for col in ckd_features_to_transform:
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    sns.histplot(
        df_cleaned[col].dropna(),
        kde=True,
        bins=20,
        ax=axes[0]
    )
    axes[0].set_title(f"Before Transformation: {col}")
    axes[0].set_xlabel(col)

    sns.histplot(
        df_ckd_transformed[col].dropna(),
        kde=True,
        bins=20,
        ax=axes[1]
    )
    axes[1].set_title(f"After Transformation: {col}")
    axes[1].set_xlabel(col)

    plt.tight_layout()
    plt.show()
```
