# **Week-3:**
#Data Cleaning:

**Objective:** Clean the dataset by fixing inconsistent values, correcting data types, handling missing values, checking duplicates, and detecting potential outliers.

# 1. Initial Data Quality Check

**Purpose:** Record the condition of the dataset before cleaning.


```python
df_cleaned = df.copy()
print("Dataset Shape:", df_cleaned.shape)
print("Total Missing Values:", df_cleaned.isnull().sum().sum())
print("Duplicate Records:", df_cleaned.duplicated().sum())
```

# 2. Clean Categorical Values

**Purpose:** Remove unnecessary spaces and tab characters from categorical values.

```python
categorical_columns = [
    "rbc", "pc", "pcc", "ba",
    "htn", "dm", "cad",
    "appet", "pe", "ane",
    "classification"
]

for col in categorical_columns:
    df_cleaned[col] = (
        df_cleaned[col]
        .astype("string")
        .str.strip()
        .str.lower()
    )
```

# 3. Verify Categorical Values

**Purpose:** Check whether categorical values are consistently formatted.

```python
for col in categorical_columns:
    print(f"\n{col}:")
    print(df_cleaned[col].unique())
```

# 4. Convert Incorrect Data Types

**Purpose:** Convert numerical values stored as text into numeric data types.

```python
numeric_object_columns = ["pcv", "wc", "rc"]

for col in numeric_object_columns:
    df_cleaned[col] = pd.to_numeric(
        df_cleaned[col],
        errors="coerce"
    )

print(df_cleaned[numeric_object_columns].dtypes)
```

# 5. Check Missing Values

**Purpose:** Identify missing values after correcting data types.

```python
missing_summary = pd.DataFrame({
    "Missing Values": df_cleaned.isnull().sum(),
    "Missing Percentage": (
        df_cleaned.isnull().sum() / len(df_cleaned) * 100
    ).round(2)
})

missing_summary
```

# 6. Identify Feature Types

**Purpose:** Separate numerical and categorical features for appropriate data cleaning.

```python
numerical_cols = df_cleaned.select_dtypes(
    include=np.number
).columns.tolist()

categorical_cols = df_cleaned.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

# Remove identifier column
numerical_cols.remove("id")

print("Numerical Features:")
print(numerical_cols)

print("\nCategorical Features:")
print(categorical_cols)
```

# 7. Handle Numerical Missing Values

**Purpose:** Replace missing numerical values using the median.

**Reason:** The median is less affected by extreme values.

```python
for col in numerical_cols:
    if df_cleaned[col].isnull().sum() > 0:
        df_cleaned[col] = df_cleaned[col].fillna(
            df_cleaned[col].median()
        )
```

# 8. Handle Categorical Missing Values

**Purpose:** Replace missing categorical values using the mode.

**Reason:** The mode represents the most frequently occurring category.

```python
for col in categorical_cols:
    if df_cleaned[col].isnull().sum() > 0:
        df_cleaned[col] = df_cleaned[col].fillna(
            df_cleaned[col].mode()[0]
        )
```

# 9. Verify Missing Values

**Purpose:** Confirm that missing values have been handled successfully.

```python
print("Total Missing Values:")
print(df_cleaned.isnull().sum().sum())

print("\nMissing Values by Column:")
print(df_cleaned.isnull().sum())
```

# 10. Update Feature Lists

**Purpose:** Confirm the numerical and categorical features before outlier detection.

```python
numerical_cols = df_cleaned.select_dtypes(
    include=np.number
).columns.tolist()

categorical_cols = df_cleaned.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

numerical_cols.remove("id")

print("Numerical Features:")
print(numerical_cols)

print("\nCategorical Features:")
print(categorical_cols)

df_cleaned.to_csv("cleaned_dataset.csv",index=False)
```

# 11. Outlier Detection Using Boxplots

**Purpose:** Visually identify potential outliers in numerical clinical features.

```python
for col in numerical_cols:
    plt.figure(figsize=(8, 4))

    df_cleaned.boxplot(column=col)

    plt.title(f"Boxplot of {col}")
    plt.ylabel(col)

    plt.show()
```

# 12. Outlier Detection Using Z-Score

**Purpose:** Identify observations that significantly deviate from the mean.

A Z-score greater than 3 or less than -3 is considered a potential outlier.

```python
all_numerical_cols = numerical_cols

z_score_df = pd.DataFrame(index=df_cleaned.index)

for col in all_numerical_cols:
    mean = df_cleaned[col].mean()
    std = df_cleaned[col].std()

    z_score_df[col] = (
        df_cleaned[col] - mean
    ) / std

z_outlier_mask = z_score_df.abs() > 3

z_outlier_report = pd.DataFrame({
    "Outlier Count": z_outlier_mask.sum(),
    "Outlier Percentage": (
        z_outlier_mask.sum() / len(df_cleaned)
    ) * 100
})

z_outlier_report
```
